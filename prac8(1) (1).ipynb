```python
plt.plot(pt, c='g', marker='o', mec='y', mfc='r', linestyle='--', linewidth=2)
plt.grid()
```

```python
plt.plot(pt,c='g', marker='x', mec='y', mfc='r')
plt.plot(pt1,c='b', marker='o', mec='y', mfc='r')
plt.xlabel("Roll_No")
plt.ylabel("Marks")
plt.title("Roll No & Marks")
```

```python
plt.plot(pt , c='b' , marker='o' , mec='r', ls=':')
plt.plot(pt1 , c='r', marker='*',mec='b', ls=':')
plt.xlabel("sample point")
plt.ylabel("values")
plt.title("point vs value")
```

```python
plt.plot(pt , c='b' , marker='o' , mec='r', dash_capstyle ='round')
plt.plot(pt1 , c='r', marker='*',mec='b', ls=':')
plt.xlabel("sample point")
plt.ylabel("values")
plt.title("point vs value")
plt.legend(['cns','dbms'])
plt.show()
```
