```python
text = """Tokenization is the first step in text analytics.
The process of breaking down a text paragraph into smaller
chunks such as words or sentences is called Tokenization."""
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print(sentences)
```

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(text)
print(words)
```

```python
import re
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
text_clean = re.sub('[^a-zA-Z]', ' ', text)
tokens = word_tokenize(text_clean.lower())
filtered = []
for w in tokens:
    if w not in stop_words:
        filtered.append(w)
print(filtered)
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

for w in filtered:
    print(ps.stem(w))
```

```python
from nltk.stem import WordNetLemmatizer

lemma = WordNetLemmatizer()

for w in filtered:
    print(lemma.lemmatize(w))
```
